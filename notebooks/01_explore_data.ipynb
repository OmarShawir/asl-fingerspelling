# Inspect a landmark clip
Run from the project root or notebooks/. The public example demonstrates the file format. To inspect a local take, use a file under data/clips/.

```python
from pathlib import Path
import json
import numpy as np

root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
path = root / 'data/sample/format_example.npz'
with np.load(path, allow_pickle=False) as clip:
    print(json.loads(clip['meta_json'].item()))
    print('Shape:', clip['xyz'].shape)
    print('Frame times (ms):', clip['t_ms'].tolist())
    print('Hand present:', clip['valid'].tolist())
```

```python
import csv
from collections import Counter

index = root / 'data/metadata.csv'
if index.is_file():
    with index.open(encoding='utf-8', newline='') as stream:
        rows = list(csv.DictReader(stream))
    print('Takes per person and session:', Counter((r['person'], r['session']) for r in rows))
    print('Takes by kind and label:', Counter((r['kind'], r['text']) for r in rows))
else:
    print('Record your first clip with scripts/collect.py to see local counts.')
```
